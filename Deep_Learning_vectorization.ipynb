#Vectorization

```python
import torch
import time

N = 100

A = torch.randn(N, N)
B = torch.randn(N, N)
C = torch.zeros(N, N)

start = time.time()

for i in range(N):
    for j in range(N):
        for k in range(N):
            C[i, j] += A[i, k] * B[k, j]

print("CPU Time:", time.time() - start)
```

```python
import torch
import time

N = 100

A = torch.randn(N, N, device="cuda")
B = torch.randn(N, N, device="cuda")

torch.cuda.synchronize()
start = time.time()

C = torch.matmul(A, B)

torch.cuda.synchronize()
print("GPU Time:", time.time() - start)
```
