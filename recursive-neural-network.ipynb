```python
import tensorflow as tf

# ---------------------------------------------------------
# RNN: h_t = f(W h_(t-1) + b)
# Calculate up to t = 20
# ---------------------------------------------------------

W = tf.constant([
    [0.5, 0.2],
    [0.1, 0.4]
], dtype=tf.float32)

b = tf.constant([0.1, 0.2], dtype=tf.float32)

def f(x):
    return tf.nn.tanh(x)


# Recursive calculation
def rnn(t, h):

    # Stopping condition
    if t > 20:
        return h

    # h_t = f(W h_(t-1) + b)
    h = f(tf.linalg.matvec(W, h) + b)

    print(f"h_{t} =", h.numpy())

    return rnn(t + 1, h)


# Initial hidden state
h = tf.constant([0.5, 0.5], dtype=tf.float32)

print("Initial hidden state:")
print("h_0 =", h.numpy())

print("\nRecursive calculation:\n")

# Start from t = 1 and continue until t = 20
h20 = rnn(1, h)

print("\nFinal hidden state:")
print("h_20 =", h20.numpy())
```
